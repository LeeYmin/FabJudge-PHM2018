# 01 — PHM2018 data structure discovery

Purpose: inspect the dataset before deciding the analysis unit, preprocessing, windows, or labels.
Put the PHM2018 files under `data/raw/` first.


```python
from pathlib import Path
import pandas as pd

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
raw_dir = project_root / 'data' / 'raw'

files = [p for p in raw_dir.rglob('*') if p.is_file() and p.name != '.gitkeep']
print(f'Found {len(files)} files')
for p in files[:50]:
    print(p.relative_to(project_root))
```

```python
# Safe first-pass inspection for CSV-like files; no preprocessing assumptions yet.
csv_files = [p for p in files if p.suffix.lower() in {'.csv', '.txt'}]

if not csv_files:
    print('No CSV/TXT files found yet.')
else:
    sample = csv_files[0]
    print('Sample file:', sample.relative_to(project_root))
    try:
        df = pd.read_csv(sample, nrows=10)
        display(df)
        print('Columns:', list(df.columns))
        print('dtypes:')
        print(df.dtypes)
    except Exception as e:
        print(type(e).__name__, e)
        print('Do not guess the delimiter/format; inspect the file format next.')
```
